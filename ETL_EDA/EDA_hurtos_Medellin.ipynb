```python
aspect_to_test = {
    'Estado civil y sexo': ('estado_civil', 'sexo'),
    'Medio de transporte y sexo': ('medio_transporte', 'sexo'),
    'Modalidad y sexo': ('modalidad', 'sexo'),
    'Arma y sexo': ('arma_medio', 'sexo'),
    'Bien y sexo': ('bien', 'sexo')
    }

for aspect, columns in aspect_to_test.items():
  print(aspect, columns)
  columna_0 = columns[0]
  columna_1 = columns[1]
  contingency_table = pd.crosstab(df_hurto_personas[columna_0], df_hurto_personas[columna_1])
  #print(contingency_table)
  chi2, p, dof, expected = chi2_contingency(contingency_table)
  significant = p < 0.05
  print(f'El valor chi-square es: {round(chi2, 3)}, el p-values es igual a {round(p, 3)}, grado de libertad: {dof}, y la significancia es: {significant}')
```
